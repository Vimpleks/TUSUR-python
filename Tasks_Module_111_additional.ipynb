```python
%%sql
CREATE TABLE product
(
    id INTEGER GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    name VARCHAR(250) NOT NULL,
    description TEXT,
    price DECIMAL(7, 2), NOT NULL
);
```

```python
%%sql
CREATE TABLE customer
(
    id INTEGER GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    ful_name VARCHAR(250) NOT NULL,
    phone VARCHAR(11), NOT NULL,
    address VARCHAR(300)
);
```

```python
%%sql
CREATE TABLE order
(
    id INTEGER GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    customer_id,
    product_id,
    quantity,
    date_order
);
```
